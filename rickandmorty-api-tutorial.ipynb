```python
import requests
import pandas as pd
```

```python
baseurl = "https://rickandmortyapi.com/api/"

endpoint = 'character'

r = requests.get(baseurl + endpoint)

# Check status code
print(r)
```

# Interrogating JSON data
After converting the response to JSON, the data is stored as a nested combination of dictionaries and lists.<br>
<br>
Values can be acessed by:<br>
- using keys for dictionaries<br>
- using index positions for lists<br>
<br>
This allows exploration of the data to extract specific information.

```python
data = r.json()

pages = data['info']['pages']
name = data['results'][0]['name']
episodes = data['results'][0]['episode']

print(len(episodes)) # count of how many episodes this character appears
```

```python
def main_request(baseurl, endpoint, x):
    r = requests.get(baseurl + endpoint, f'?page={x}')
    return r.json()

def get_pages(response):
    pages = response['info']['pages']
    return pages

def parse_json(response):
    character_list = []
    for item in response['results']:
        character = {
            'id': item['id'],
            'name': item['name'],
            'number_of_episodes': len(item['episode']),
        }

        character_list.append(character)

    return character_list

    # name = data['results'][0]['name']
    # episodes = data['results'][0]['episode']

mainlist = []
data = main_request(baseurl, endpoint, 8)
for x in range(1,get_pages(data)+1):
    print(x)
    mainlist.extend(parse_json(main_request(baseurl, endpoint, x)))

df = pd.DataFrame(mainlist)

df.to_csv('CharacterList.csv', index=False)



```
