```python
import requests
```

```python
# Create a variable to store the data retrieved from the API
response = requests.get("http://randomfox.ca/floof")
```

### Access the status code of request

A `status code = 200` means successful.

```python
# Access the status code of request
print(response.status_code)
```

### Working with API data

- `.text` returns the API response as plain text (a string).
- `.json()` transforms that data into a structured Python object (dict)

```python
# Store the parsed JSON response as a Python dictionary
fox = response.json()

# Access a value from the dictionary using bracket notation:
print(fox["image"])
```
